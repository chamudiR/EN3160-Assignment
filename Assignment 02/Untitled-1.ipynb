```python
import numpy as np
import cv2
from scipy import ndimage
from skimage.feature import peak_local_max  # Import for local maxima detection

# Optional: Set a random seed for reproducibility (not required for blob detection)
# np.random.seed(0)

def laplacian_of_gaussian(image, sigma):
    ksize = int(6*sigma + 1) | 1  # ensure odd kernel size
    log_kernel = cv2.getGaussianKernel(ksize, sigma) * cv2.getGaussianKernel(ksize, sigma).T
    log_kernel = log_kernel - log_kernel.mean()
    response = cv2.filter2D(image, cv2.CV_64F, log_kernel)
    return response

# Load the image in grayscale
image = cv2.imread(r'D:\5th sem\Image proccessing\EN3160-Assignment-1---Intensity-Transformations-and-Neighborhood-Filtering\Assignment 02\the_berry_farms_sunflower_field.jpeg', cv2.IMREAD_GRAYSCALE)
sigmas = np.linspace(2, 12, 13)
h, w = image.shape
stack = np.zeros((h, w, len(sigmas)))

# Build scale-space
for i, sigma in enumerate(sigmas):
    log_resp = ndimage.gaussian_laplace(image, sigma)
    stack[:, :, i] = sigma**2 * np.abs(log_resp)  # scale normalization

# Detect local maxima in 3D space (x, y, sigma)
# Set a reasonable threshold (adjust as needed)
threshold = 0.03
coordinates = peak_local_max(stack, threshold_abs=threshold, footprint=np.ones((3, 3, 3)))
blobs = [(y, x, sigmas[z]) for y, x, z in coordinates]

# Optional: Print or visualize the detected blobs
print("Number of blobs detected:", len(blobs))
for blob in blobs:
    print(f"y: {blob[0]}, x: {blob[1]}, sigma: {blob[2]}")

```
